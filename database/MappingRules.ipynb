<h1>Mapping Rules: Theory</h1>

```python
import pandas as pd
```

## Rule 1 &mdash; Mapping Strong Entity Sets

We keep strong entity sets as separate relations because they already represent
**independelty identfiable objects**, and their attributes belong to those objects.

```python


df_users = pd.DataFrame({
    "UserID": [1, 2, 3, 4],
    "Username": ["alice", "bob", "carol", "david"],
    "Email": [
        "alice@example.com",
        "bob@example.com",
        "carol@example.com",
        "david@example.com"
    ]
})

df_users.set_index("UserID", inplace=True)
df_users
```

Note that both `userID` and `Email` would serve as a primary key.


## Rule 2 &mdash; Mapping Weak Entity Sets

Suppose an exercise can contain several subquestions. 

We have two entity sets:

1. `Exercise`: A strong entity set, uniquely identified by `ExerciseID`;
2. Exercise_Part: A weak entity set, identified by `PartNumber` within its own
exercise.

For example, Exercise 101 has parts 1, 2, and 3. Exercise 102 also has parts 1,
and 2.

The important point is that `PartNumber` is not globally unique. Its value
only has meaning within a particular exercise.



```python


df_exercises = pd.DataFrame({
    "ExerciseID": [101, 102, 103],
    "Title": [
        "SQL Basics",
        "Normalization",
        "ER Modeling"
    ]
})

df_exercise_parts = pd.DataFrame({
    "ExerciseID": [101, 101, 101, 102, 102, 103],
    "PartNumber": [1, 2, 3, 1, 2, 1],
    "Question": [
        "Define a primary key.",
        "Define a foreign key.",
        "Explain a JOIN.",
        "Explain 1NF.",
        "Explain 2NF.",
        "Define a strong entity."
    ]
})
```

Note that the PartNumber alone is not unique in the weak entity:

```python
df_exercise_parts['PartNumber'].is_unique
```

```python
# ------------------------------------------------------------------------------
# Is the combination of the PartNumber and ExerciseID unique?
# ------------------------------------------------------------------------------
not df_exercise_parts.duplicated(subset=['PartNumber', 'ExerciseID']).any()
```

Suppose we would add the new part to the `Exericise_Parts` relation:


```python
new_part = {
    "ExerciseID": 101,
    "PartNumber": 2,
    "Question": "What is a candidate key?"
}
```
