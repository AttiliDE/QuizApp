<h2>The Entity Relationship Model</h2>

### The Basic Parts of an ER-Model

An ER-Model consists of three parts:

1. **Entity-sets:**
   - An **entity** is an object in the real world and entities are
     distinguishable from all other objects (e.g. a person, a house).
   - An **entity set** is a set of entities of the **same type**
     that share the same **properties**, or **attributes**.

2. **Attributes:**
   - The set of properties that describes an entity.

3. **Relationship-sets:**
   - These are associations among several entities.
   - A **relationship set** is a **set of relationships** of the
     **same type**.
   - $R\subseteq \{\left(e_1, e_2, \dots, e_n\right) | e_{1} \in E_{1}, \in E_{2},
   \dots, e_n \in E_n\}$ is a <strong>relationship</strong>.
   - The **part** an **entity** plays in a **relationship** is the **entity's role**.

### Mapping Cardinalities

- Cardinalities are used to express the number of **entities** to which **another
    entity** can be **associated** via **relationship set**.
- Most useful in describing **binary relationship sets**.

- **Cardinalities:**
  - Cardinalities express the number of **entities** to which
    **another entity** can be associated via a **relationship set**.
  - They are particularly useful for describing **binary relationship sets**.
  - The notation distinguishes:
    - **1:** Exactly one entity.
    - **c:** Zero or one entity, where $c \in \{0,1\}$.
    - **m, n:** Multiple entities (with the possibility of zero).

- **Types of Cardinalities:**

  1. **One-to-One (1:1):**
     - Each entity of entity set A is associated with exactly one
       entity of entity set B.
     - Each entity of entity set B is associated with exactly one
       entity of entity set A.

  2. **One-to-Optional-One (1:c):**
     - Each entity of entity set A is associated with zero or one
       entity of entity set B.
     - Each entity of entity set B is associated with exactly one
       entity of entity set A.

  3. **Optional-One-to-Optional-One (c:c):**
     - Each entity of entity set A is associated with zero or one
       entity of entity set B.
     - Each entity of entity set B is associated with zero or one
       entity of entity set A.

  4. **Many-to-One (m:1):**
     - Each entity of entity set A can be associated with multiple
       entities of entity set B.
     - Each entity of entity set B is associated with exactly one
       entity of entity set A.

  5. **Many-to-Optional-One (m:c):**
     - Each entity of entity set A can be associated with multiple
       entities of entity set B.
     - Each entity of entity set B is associated with zero or one
       entity of entity set A.

  6. **Many-to-Many (m:n):**
     - Each entity of entity set A can be associated with multiple
       entities of entity set B.
     - Each entity of entity set B can be associated with multiple
       entities of entity set A.

### Types of Entity and Relationship Sets

An entity set can be either strong or weak.

- **Strong entity set**: A strong entity set is an entity set that possesses its own primary key, allowing each entity to be uniquely identified independently of other entity sets.
- **Weak entity set**: A weak entity set does not possess a sufficient key of its own and depends on an identifying strong (or owner) entity set for its identification.

Whenever a relationship is formed between two strong entity sets, it is itself called 
an **strong relationship**.

### The Seven Rules of ER-To-Relational Mapping

**ER-to-relational mapping** is the process of transforming an Entity-Relationship (ER)
model into a realtional database schema. It defines how entity sets, attributes,
and relationship sets are represented as tables, columns, primary keys, and 
foreign keys.

1. **Mapping Strong Entity Sets**:

    For each strong entity set, create a relation containing all its simple attibutes.
    - Choose one candidate key as the **primary key** of the relation.
    - Composite attributes are replaced by their simple components.
    - Derived attributes are generally not stored.

    $\to$ A strong entity set can be uniquely identified by its own attributes, independently
    of other entity sets.


2. **Mapping Weak Entity Sets**

    For each weak entity set, create a separate relation containing its attributes
    and the primary key of its identifying owner entity set.
    - The primary key of the owner becomes a **foreign key** in the weak entity's
    relation.
    - The primary key of the weak entity's relation is normally composed of the 
    owner's pimary key and the weak entity's **partial key**.
    - The identifying foreign key must be declared **NOT NULL**, since a weak
    entity cannot exist without its owner.
    - The foreign key must include `ON DELETE CASCADE` and `ON UPDATE CASCADE`
    to maintain referential integrity.
    - Attributes of the identifying relationship can also be included in the weak
    entity's relation.

    $\to$ The resulting composite primary key uniquely identifies each weak
    entity within the context of its owner.

3. **Mapping Binary Many-to-Many (M:N) Relationship Sets**

    For each binary M:N relationship set, create a separate relation, called
    a **relationship relation**, representing the associations between the 
    participating entity sets.

    - The primary keys of both participating entity sets are included in the
    new relation as **foreign keys**.
    - These two foreign keys together form the **composite primary key** of the
    relationship relation.
    - Any attributes belonging to the relationship set are included in the new
    relation.
    - The original M:N relationship is represented through two 1:N relationships
    involving the new relation.

    $\to$ A separate relationship relation is necessary because neither
    participating entity can generally represent multiple associations using
    a single foreign key.

4. **Mapping Binary Many-to-One (M:1) Relationship Sets**

    For each binary M:1 relationship set, include the primary key of the entity
    set on the **1-side** as a foreign key in the relation on the **M-side**.
    - The foreign key references the primary key of the entity set on the 1-side.
    - The foreign key must be declared `NOT NULL`, since every entity on the
    M-side must be associated with exactly one entity on the 1-side.
    - The foreign key does not require a `UNIQUE` constraint, since multiple
    entities on the M-side may reference the same entity on the 1-side.
    - Attributes belonging to the relationship set can be included in the relation
    on the M-side.

    $\to$ A separate relationship rerlation is generally unnecessary because
    the association can be represented by a foreign key on the M-side.

5. **Mapping Binary Many-to-Optional-One (M:c) Relationship Set**

    For each binary M:c ($c\in \{0, 1\}$) relationship set, include the primary
    key of the entity set on the optional-side as a foreign key in the relation
    on the many-side.

    - The foreign key references the primary key of the entity set on the
    optional-one side.
    - The foreign key does not require a **NOT NULL** constraint, since an 
    entity on the many-side may exist without being associated with an entity
    on the optional-one side.
    - The foreign key does not require a `UNIQUE` constraint, since multiple
    entities may reference the same entity.
    - Attributes belonging to the relationship set can be included in the relation
    containing the foreign key.

    $\to$ This mapping is similar to M:1, but the foreign key is allowed to 
    contain `NULL` values because participation is optional.

6. **Mapping Binary One-to-One Relationships sets**

    For each binary 1:1 relationship, introduce a foreign key into one of the
    participating relations.
    - The foreign key references the primary key of the other relation.
    - The foreign key must be `UNIQUE` to ensure that each entity is associated
    with at most one entity of the other set.
    - The foreign key must be declared `NOT NULL`, since participation in 1:1
    relationship is mandatory.
    - Attributes belonging to the relationship set can be included in the relation
    containing the foreign key.
    - A separate relationship relation is generally unnecessary.
    - Prefer placing it in the rerlation with **total participation**, when
    appropriate.
    - Any attributes belonging to the relationship can be included in the same 
    relation.

    $\to$ However, UNIQUE and NOT NULL only guarantee that every entity in
   the relation containing the foreign key references exactly one
   entity in the other relation. They do not guarantee that every
   entity in the referenced relation participates in the relationship.

   Therefore, enforcing mandatory participation on both sides may
   require additional database constraints or a different schema design.

7. **Mapping Binary Optional-One-to-One (c:1) Relationship Sets**

   For each binary c:1 relationship set, introduce a foreign key
   into the relation on the **c-side**, referencing the primary
   key of the entity set on the 1-side.

   - The foreign key references the primary key of the entity set
     on the 1-side.
   - The foreign key must be declared `NOT NULL`, since every
     entity on the c-side must be associated with exactly one
     entity on the 1-side.
   - The foreign key must be declared `UNIQUE`, since an entity
     on the 1-side can be associated with at most one entity
     on the c-side.
   - Attributes belonging to the relationship set can be included
     in the relation containing the foreign key.

   $\to$ The combination of NOT NULL and UNIQUE ensures that every entity
   on the c-side references exactly one entity on the 1-side,
   while preventing multiple entities from referencing the same entity.
   
   A separate relationship relation is generally unnecessary.

8. **Mapping Binary Optional-One-to-Optional-One (c:c) Relationship Sets**

   For each binary c:c relationship set, introduce a foreign key
   into one of the two participating relations.
   - The foreign key references the primary key of the other relation.
   - The foreign key must be declared `UNIQUE`, since each entity
     can be associated with at most one entity of the other set.
   - The foreign key does not require a `NOT NULL` constraint,
     since participation is optional.
   - The foreign key can be placed in either participating relation,
     depending on the database design.
   - Attributes belonging to the relationship set can be included
     in the relation containing the foreign key.

   $\to$ A separate relationship relation is generally unnecessary because
   the association can be represented by a nullable, unique foreign key.
